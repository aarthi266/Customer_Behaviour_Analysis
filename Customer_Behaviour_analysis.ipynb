```python
data.columns=data.columns.str.lower()
data.columns=data.columns.str.replace(' ','_')
data=data.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
data.columns
```

#Create a column age_group

```python
labels=['Young Adult','Adult','Middle-aged','Senior']
data['age_group']=pd.qcut(data['age'],q=4,labels=labels)
```

```python
data[['age_group','age']].head()
```

Create Purchase Columns purchase_frequency_days

```python
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90}
data['purchase_frequency_days']=data['frequency_of_purchases'].map(frequency_mapping)
```
